```python
# Numerical features
numerical_features = [
    'LeadTime',
    'Adults',
    'Children',
    'Babies',
    'TotalOfSpecialRequests',
    'StaysInWeekendNights',
    'StaysInWeekNights'
]

# Cyclic date features
date_features = [
    'Month_sin',
    'Month_cos'
]

# One-hot encoded features
encoded_features = [
    col for col in cc.columns
    if col.startswith((
        'Meal_',
        'Country_',
        'MarketSegment_',
        'DistributionChannel_',
        'ReservedRoomType_',
        'CustomerType_'
    ))
]

# Create a separate DataFrame for K-Means
selected_features = numerical_features + date_features + encoded_features

kmeans_data = cc[selected_features].copy()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

numerical_features = [
    'LeadTime',
    'Adults',
    'Children',
    'Babies',
    'TotalOfSpecialRequests',
    'StaysInWeekendNights',
    'StaysInWeekNights'
]

plt.figure(figsize=(14, 8))

for i, col in enumerate(numerical_features, 1):
    plt.subplot(3, 3, i)
    sns.boxplot(y=kmeans_data[col])
    plt.title(col)

plt.tight_layout()
plt.show()
```

```python
for col in numerical_features:

    Q1 = kmeans_data[col].quantile(0.25)
    Q3 = kmeans_data[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = kmeans_data[
        (kmeans_data[col] < lower) |
        (kmeans_data[col] > upper)
    ]

    print(f"\n{col}")
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower boundary:", lower)
    print("Upper boundary:", upper)
    print("Outlier count:", len(outliers))
    print("Outlier percentage:", round(len(outliers) / len(kmeans_data) * 100, 2), "%")
```
